# Section 2: Data Wrangling

To prepare out data for analysis, we need to perform data wrangling. In this section, we will learn how to clean and transform our data to make it suitable for analysis.


### 🐧 Load the raw dataset

````python
import pandas as pd

# Load raw data
url = "https://raw.githubusercontent.com/allisonhorst/palmerpenguins/master/inst/extdata/penguins_raw.csv"
penguins = pd.read_csv(url)

penguins.head()
````

### 🧼 Step 1: Drop unneeded columns

````python
# Check columns
penguins.columns
````

````python
# Drop columns not in cleaned version (e.g. 'studyName', 'Individual ID', 'Comments')
penguins = penguins.drop(columns=['studyName', 'Sample Number', 'Region', 'Stage',
                                  'Individual ID', 'Clutch Completion', 'Delta 15 N (o/oo)', 'Delta 13 C (o/oo)', 'Comments'])

penguins.head()
````



➤ *Why drop columns?*  
Columns irrelevant to the analysis or full of missing/inconsistent data are noise. Drop them early.

### 🏷️ Step 2: Rename columns to match cleaned format
Clean column names to snake_case and consistent naming

````python

penguins = penguins.rename(columns={
    'Species': 'species',
    'Island': 'island',
    'Sex': 'sex',
    'Beak Length (mm)': 'bill_length_mm',
    'Beak Depth (mm)': 'bill_depth_mm',
    'Flipper Length (mm)': 'flipper_length_mm',
    'Body Mass (g)': 'body_mass_g',
    'Date Egg': 'year'
})

penguins.head()
````

### 🔁 Step 3: Convert columns to the correct data types and manage the content of the columns

````python
penguins.dtypes
````

Change the content of the species column to just the name of the penguing species and make it as a category type.

````python
penguins.species.unique()
````

````python
# Just the the first word of the species column
penguins['species'] = penguins['species'].str.split().str[0]

# Convert species to categorical
penguins['species'] = penguins['species'].astype('category')
penguins.head()
````

Convert the column with the date in the format of the date and keep just the year

````python
penguins['year'] = pd.to_datetime(penguins['year'])
penguins.head()
````

````python
penguins.dtypes
````

````python
# Keep only the year part
penguins['year'] = penguins['year'].dt.year
penguins.head()
````

````python
penguins['year'].dtypes
````

````python
# Convert sex to lower case and make it a category type
penguins['sex'] = penguins['sex'].str.lower().astype('category')

# Check types
penguins.dtypes
````

➤ *Why care about types?*  
Performance, memory, and avoiding weird bugs when modeling or grouping.

## 📦 GroupBy — Split, Apply, Combine

````python
url_penguins = "https://raw.githubusercontent.com/allisonhorst/palmerpenguins/master/inst/extdata/penguins.csv"
penguins = pd.read_csv(url_penguins)
penguins.head()
````

### Aggregating

````python
penguins['body_mass_g'].mean()
````

````python
species_map = pd.Series([], dtype=float)

for species in penguins['species'].unique():
    species_only = penguins.loc[penguins['species'] == species]
    species_map.loc[species] = species_only['body_mass_g'].mean()

species_map
````

### Grouping

````python
# Grouping the 'penguins' DataFrame by 'species' and calculating the mean of 'body_mass_g'
penguins.groupby('species')['body_mass_g'].mean()
````

### ➕ Multiple Columns Aggregated

````python

penguins.groupby('species')[['body_mass_g', 'flipper_length_mm']].mean()
````

### 🧠 Grouping by Multiple Keys

````python
penguins.groupby(['species', 'sex'])['body_mass_g'].mean()
````

### 🧮 Use `.agg()` for Multiple Functions

````python

penguins.groupby('species')['body_mass_g'].agg(['mean', 'std', 'count'])
````

### 🧪 Group All Rows, No Column Specified

````python

penguins.groupby('species').mean(numeric_only=True)
````

### 🧰 Reset Index for Flattened Output

````python
penguins.groupby('species')['body_mass_g'].mean().index #goupby usa come indici i valori utilizzati per fare lo split
````

````python

penguins.groupby('species')['body_mass_g'].mean().reset_index() #alcune funzioni necessitano invece di indici numerici
````

## 🧭 Penguin Pivot Tables — From Flat to Insightful

````python
# Import clean data
url_penguins = "https://raw.githubusercontent.com/allisonhorst/palmerpenguins/master/inst/extdata/penguins.csv"
penguins = pd.read_csv(url_penguins)
penguins.head()
````

### 💡 What is a Pivot Table?

```python
pd.pivot_table(df,
index = cols, #  keys to group by for index  
columns = cols2, #  keys to group by for columns
values = cols3, # columns to aggregate
aggfunc = func # aggregation function
)
```

### 📌 Basic Example: Average Body Mass by Species and Sex

Compare **average body mass** per **species**, broken down by **sex**.

````python
pd.pivot_table(
    penguins,
    index='species',
    columns='sex',
    values='body_mass_g',
    aggfunc='mean'
)
````

### 🔁 Change the Aggregation Function

You're not limited to means. Use `aggfunc` to calculate other statistics.

````python
pd.pivot_table(
    penguins,
    values='flipper_length_mm',
    index='species',
    columns='sex',
    aggfunc='median'
)
````

### 🔢 Multiple Aggregated Values

You can calculate multiple statistics at once.

````python
pd.pivot_table(
    penguins,
    values=['body_mass_g', 'flipper_length_mm'],
    index='species',
    columns='sex',
    aggfunc='mean'
)
````

### 📊 Count Instead of Mean

````python
pd.pivot_table(
    penguins,
    index='species',
    columns='sex',
    values='body_mass_g',
    aggfunc='count'
)
````

````python
pd.pivot_table(
    penguins,
    index='species',
    columns='sex',
    aggfunc='size'
)
````

### 🧪 Using a `lambda` in a Pivot Table

#### 💡 Goal: Compare the *range* of body mass (max - min) per species and sex

````python

pd.pivot_table(
    penguins,
    values='body_mass_g',
    index='species',
    columns='sex',
    aggfunc=lambda x: x.max() - x.min()
)
````
